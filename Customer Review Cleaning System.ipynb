```python
import numpy as np
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# Data Preprocessing
data = pd.read_csv("IMDB Dataset.csv", nrows=5000)

print(data.head())
print(data.shape)
data.info()

print("\nTotal Number of Missing Values")
print("----------------------------------")
print(data.isnull().sum())

print("\nTotal Duplicates:", data.duplicated().sum())

print(
    "\nTotal Duplicates in Review:",
    data.duplicated(subset=['review']).sum()
)

# Remove duplicate reviews
data = data.drop_duplicates(subset=['review'])

# Stopwords and Stemming
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()

# Cleaning Function
def clean(text):

    # Convert to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', ' ', text)

    # Remove punctuation, numbers and special characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenization
    words = word_tokenize(text)

    # Remove stopwords
    words = [
        word for word in words
        if word not in stop_words
    ]

    # Stemming
    words = [
        stemmer.stem(word)
        for word in words
    ]

    # Convert words back to text
    return " ".join(words)

# Apply Cleaning
data["cleaned_review"] = data["review"].apply(clean)

# Compare Original and Cleaned
print("\nOriginal vs Cleaned Reviews:")
print("----------------------------------")

print(
    data[["review", "cleaned_review", "sentiment"]].head(10)
)

# Export Dataset

data.to_csv(
    "IMDB_Preprocessed.csv",
    index=False
)

print("\nPreprocessed dataset saved successfully!")
```
